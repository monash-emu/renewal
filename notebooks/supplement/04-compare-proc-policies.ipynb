```python
import json
import pandas as pd
from IPython.display import display, Markdown, Latex
import pycountry_convert as pc
from matplotlib_inline.backend_inline import set_matplotlib_formats

from emu_renewal.constants import CONT_CMAP, DATA_PATH, FULL_RUN, OXCGRT_COLMAP, OXCGRT_LOCS
from emu_renewal.plotting import compare_proc_pol
from emu_renewal.utils import get_countries_by_continent, split_list_into_segments, get_analysis_paths, get_analysis_commits_df

set_matplotlib_formats("svg")
```

```python
all_countries = json.load(open(DATA_PATH / "config/oxcgrt_included.json", "r"))
analysis_paths = get_analysis_paths(FULL_RUN, all_countries)
countries_by_cont = get_countries_by_continent(all_countries)
```

# Purpose
This document presents the residual transmission scaling (with associated uncertainty) implemented under the baseline analysis (without scaling according to policy data for each country analysed).
It is intended to provide a comparison between the residual non-mechanistic transmission that needed to be applied for each analysis in the absence of scaling and the OxCGRT time series
data, with each included policy domain weighted equally, with various combinations of policies
included in the overall metrics.
That is, this gives a naive comparison of the estimated fluctuation in transmission not
captured through the mechanistic model and the raw OxCGRT data.

# Included policies
The table below lists the OxCGRT indicators included in each composite policy map.

```python
all_codes = list(sorted({code for codes in OXCGRT_COLMAP.values() for code in codes}))
policies = [OXCGRT_LOCS[code] for code in all_codes]
included_policies = pd.DataFrame({"indicator": all_codes, "policy": policies})
for name, codes in OXCGRT_COLMAP.items():
    included_policies[name] = ["yes" if code in codes else "" for code in all_codes]
display(Markdown(included_policies.to_markdown(index=False)))
display(Latex(r"\newpage"))
```

```python
display(Markdown("# Policy profile comparisons \n\n"))
display(Latex(r"\newpage"))
for cont in CONT_CMAP:
    countries = countries_by_cont[cont]
    display(Markdown(f"### {pc.convert_continent_code_to_continent_name(cont)}"))
    country_groups = split_list_into_segments(countries, 16)
    for group in country_groups:
        display(compare_proc_pol(analysis_paths, group, 4))
```
